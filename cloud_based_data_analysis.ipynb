```python
data = {
    "Student": ["Asha", "Rahul", "Priya", "Kavin", "Meena",
                "Arun", "Divya", "Vijay", "Anu", "Ravi"],

    "Department": ["CSE", "ECE", "CSE", "IT", "CSE",
                   "ECE", "IT", "CSE", "ECE", "IT"],

    "Attendance": [92, 85, 95, 78, 88, 82, 91, 96, 84, 89],

    "Marks": [88, 76, 92, 68, 85, 72, 90, 95, 79, 86]
}

df = pd.DataFrame(data)

df
```

```python
dataset_path = project_folder + "/student_data.csv"

df.to_csv(dataset_path, index=False)

print("Dataset saved to Google Drive!")
print(dataset_path)
```

```python
df = pd.read_csv(dataset_path)

print("Dataset loaded from Google Drive!")
df
```

```python
print("Total Students:", len(df))
```

```python
print("Average Marks:", df["Marks"].mean())
```

```python
print("Average Attendance:", df["Attendance"].mean())
```

```python
print("Highest Marks:", df["Marks"].max())
```

```python
print("Lowest Marks:", df["Marks"].min())
```

```python
department_analysis = df.groupby("Department")["Marks"].mean()

print(department_analysis)
```

```python
department_analysis.plot(kind="bar")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.bar(df["Student"], df["Attendance"])

plt.title("Student Attendance")
plt.xlabel("Student")
plt.ylabel("Attendance (%)")

plt.xticks(rotation=45)

plt.show()
```

```python
top_students = df.sort_values(
    by="Marks",
    ascending=False
).head(5)

print("Top 5 Students:")
top_students
```

```python
def performance_category(marks):
    if marks >= 90:
        return "Excellent"
    elif marks >= 75:
        return "Good"
    elif marks >= 60:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Marks"].apply(performance_category)

df
```

```python
summary = {
    "Total Students": len(df),
    "Average Marks": df["Marks"].mean(),
    "Average Attendance": df["Attendance"].mean(),
    "Highest Marks": df["Marks"].max(),
    "Lowest Marks": df["Marks"].min()
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

summary_df
```

```python
report_path = project_folder + "/analysis_report.csv"

summary_df.to_csv(report_path, index=False)

print("Analysis report saved to Google Drive!")
```

```python
from google.colab import files
import pandas as pd

print("Upload your CSV dataset")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\nDataset uploaded successfully!")
print("File:", file_name)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("===== DATASET INFORMATION =====")

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(list(df.columns))

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing numerical values with the mean
numeric_columns = df.select_dtypes(include='number').columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("Data cleaning completed!")

df.head()
```

```python
print("===== STATISTICAL SUMMARY =====")

df.describe()
```

```python
numeric_columns = df.select_dtypes(include='number').columns

print("Numerical columns:", list(numeric_columns))
```

```python
for column in numeric_columns:
    plt.figure(figsize=(8, 4))

    plt.hist(df[column], bins=10)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()
```
